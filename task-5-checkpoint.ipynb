```python

# # TASK 5: Classification Tasks Overview

# ## Introduction
# 1.  **Student Pass/Fail Prediction:** A binary classification task using numerical data.
# 2.  **Sentiment Analysis:** A binary classification task using natural language processing (NLP).
import pandas as pd
import numpy as np
import io
# Visualization
import matplotlib.pyplot as plt
import seaborn as sns
# Machine Learning - Task 1 (Student Pass/Fail)
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
# Machine Learning - Task 2 (Sentiment Analysis)
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import RegexpTokenizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import precision_score, recall_score, f1_score, classification_report
import nltk

# Download necessary NLTK data (only need to run this once)
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')

# Set plot style for visualizations
plt.style.use('seaborn-v0_8-whitegrid')
# ---
# ## Task 1: Student Pass/Fail Prediction
# ---
# **Objective:** Predict whether a student will pass or fail based on their study hours and attendance.
# ### 1. Dataset Selection (Create a Dataset)
# Generate synthetic data
np.random.seed(42) # for reproducibility
num_students = 150

# Generate features: Study Hours and Attendance
study_hours = np.random.uniform(1, 20, num_students)
attendance = np.random.uniform(50, 100, num_students)

# Generate the target variable 'Pass' based on a logical rule with some noise
# Rule: Higher study hours and attendance increase the probability of passing
probability_of_passing = ((study_hours / 20) + (attendance / 100)) / 2
# Add some randomness so the model is not perfect
probability_of_passing += np.random.normal(0, 0.1, num_students)

# Students with a probability > 0.6 are classified as 'Pass' (1)
passed = (probability_of_passing > 0.6).astype(int)

# Create the DataFrame
df_students = pd.DataFrame({
    'Study_Hours': study_hours,
    'Attendance': attendance,
    'Pass': passed
})

print("Generated Student Dataset (First 5 Rows):")
print(df_students.head())

print("\nNumber of students who passed vs. failed:")
print(df_students['Pass'].value_counts())

# ### 2. Tasks to Perform
# #### a. Data Exploration
# Check for missing values
print("\nMissing Values:")
print(df_students.isnull().sum())

# Basic stats and outlier check
print("\nSummary Statistics:")
print(df_students.describe())

# Plot the relationship to visualize trends
plt.figure(figsize=(10, 7))
sns.scatterplot(data=df_students, x='Study_Hours', y='Attendance', hue='Pass', style='Pass', s=100)
plt.title('Student Pass/Fail Status based on Study Hours and Attendance')
plt.xlabel('Weekly Study Hours')
plt.ylabel('Class Attendance (%)')
plt.legend(title='Status', labels=['Fail (0)', 'Pass (1)'])
plt.show()

# **Exploration Insights:**
# The scatter plot clearly shows a trend. Students who pass (blue circles) tend to be concentrated in the upper-right section, indicating higher study hours and attendance. Students who fail (orange crosses) are mostly in the lower-left. This visual separation suggests that a classification model should perform well.
# #### b. Model Training
# - Train a Logistic Regression model.
# Define features (X) and target (y)
X_students = df_students[['Study_Hours', 'Attendance']]
y_students = df_students['Pass']

# Split the data into training and testing sets (80-20 split)
X_train_stud, X_test_stud, y_train_stud, y_test_stud = train_test_split(
    X_students, y_students, test_size=0.2, random_state=42, stratify=y_students
)

# Initialize and train the Logistic Regression model
log_reg_model = LogisticRegression(random_state=42)
log_reg_model.fit(X_train_stud, y_train_stud)

print("Logistic Regression model trained successfully.")

# #### c. Model Evaluation
# Make predictions on the test set
y_pred_stud = log_reg_model.predict(X_test_stud)

# --- Deliverables for Task 1 ---

# 1. Classification Model (is `log_reg_model`)

# 2. Evaluation Metrics
# Calculate Accuracy
accuracy = accuracy_score(y_test_stud, y_pred_stud)
print(f"Model Accuracy: {accuracy:.2f}")

# Generate the Confusion Matrix
cm = confusion_matrix(y_test_stud, y_pred_stud)

# Visualize the Confusion Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Fail (0)', 'Pass (1)'],
            yticklabels=['Fail (0)', 'Pass (1)'])
plt.title('Confusion Matrix - Student Pass/Fail Prediction')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()

# 3. Insights
print("\n--- Insights ---")
print("The model achieved a high accuracy, indicating it is effective at predicting student outcomes from the given features.")
print("\nThe Confusion Matrix shows:")
print(f"  - True Negatives (Correctly predicted 'Fail'): {cm[0][0]}")
print(f"  - False Positives (Incorrectly predicted 'Pass'): {cm[0][1]}")
print(f"  - False Negatives (Incorrectly predicted 'Fail'): {cm[1][0]}")
print(f"  - True Positives (Correctly predicted 'Pass'): {cm[1][1]}")
print("\nKey Predictors of Student Performance: The model confirms that both 'Study_Hours' and 'Attendance' are strong predictors of whether a student will pass.")

# ---
# ## Task 2: Sentiment Analysis with Natural Language Processing
# ---
# **Objective:** Analyze customer reviews to classify sentiments as positive or negative.

# ### 1. Dataset Selection (Create a Dataset)
# We will generate a `reviews.csv` dataset.
# Sample reviews
positive_reviews = [
    "This product is absolutely wonderful, I loved it!",
    "Fantastic quality and great service. Highly recommend.",
    "A truly amazing experience from start to finish. Excellent!",
    "I am so happy with this purchase. It exceeded my expectations.",
    "Best purchase I've made all year. The quality is superb.",
    "The customer support was so helpful and friendly. A pleasure to deal with.",
    "Brilliant! It works perfectly and is easy to use.",
    "A solid 5-star product. Well done!",
    "I couldn't be happier with the result. Simply the best.",
    "Good value and works as advertised. I am very satisfied."
]

negative_reviews = [
    "This was a terrible purchase. It broke after one day.",
    "Awful product and even worse customer service. Avoid at all costs.",
    "A complete waste of money. I am very disappointed.",
    "It did not work as advertised. Very misleading information.",
    "I regret buying this. The quality is extremely poor.",
    "The item arrived damaged. What a bad experience.",
    "This is the worst product I have ever bought. So frustrating.",
    "Zero stars if I could. A total failure.",
    "I'm so angry about this purchase. A piece of junk.",
    "Do not buy this. It's a scam and a complete disappointment."
]

# Create the DataFrame
review_texts = positive_reviews + negative_reviews
sentiments = ['positive'] * len(positive_reviews) + ['negative'] * len(negative_reviews)

df_reviews = pd.DataFrame({
    'Review_Text': review_texts,
    'Sentiment': sentiments
})

# Shuffle the dataset to mix positive and negative reviews
df_reviews = df_reviews.sample(frac=1, random_state=42).reset_index(drop=True)

# Map sentiment labels to numbers (0 for negative, 1 for positive)
df_reviews['Sentiment_Label'] = df_reviews['Sentiment'].map({'negative': 0, 'positive': 1})

print("Generated Reviews Dataset:")
print(df_reviews)

# ### 2. Tasks to Perform
# #### a. Load and Preprocess the Dataset
# - Remove stopwords, punctuation, special characters.
# - Convert text to lowercase.
# - Tokenize and lemmatize the text.

lemmatizer = WordNetLemmatizer()
stopwords_eng = set(stopwords.words('english'))
tokenizer = RegexpTokenizer(r'\w+') # Matches word characters

def preprocess_text(text):
    # 1. Lowercase
    text = text.lower()
    # 2. Tokenize and remove punctuation
    tokens = tokenizer.tokenize(text)
    # 3. Remove stopwords
    filtered_tokens = [word for word in tokens if word not in stopwords_eng]
    # 4. Lemmatize
    lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]
    return " ".join(lemmatized_tokens)

# --- Deliverable 1: Preprocessed Dataset ---
df_reviews['Cleaned_Text'] = df_reviews['Review_Text'].apply(preprocess_text)

print("\nDataset with Cleaned and Tokenized Text Data:")
print(df_reviews[['Review_Text', 'Cleaned_Text']].head())
# #### b. Text Vectorization
# - Convert text to a numerical format using TF-IDF.

# Initialize the TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer(max_features=100) # Limit to top 100 features for simplicity

# Fit and transform the cleaned text
X_reviews_tfidf = tfidf_vectorizer.fit_transform(df_reviews['Cleaned_Text'])

# Define target variable
y_reviews = df_reviews['Sentiment_Label']

# #### c. Model Training & Evaluation

# Split data into training and testing sets
X_train_rev, X_test_rev, y_train_rev, y_test_rev = train_test_split(
    X_reviews_tfidf, y_reviews, test_size=0.2, random_state=42, stratify=y_reviews
)

# --- Deliverable 2: Sentiment Classification Model ---
# Initialize and train the Logistic Regression model
sentiment_model = LogisticRegression(random_state=42)
sentiment_model.fit(X_train_rev, y_train_rev)
print("Sentiment classification model trained.")

# Make predictions
y_pred_rev = sentiment_model.predict(X_test_rev)

# --- Deliverable 3: Evaluation Report ---
print("\n--- Evaluation Report ---")
# Use classification_report for a comprehensive summary
report = classification_report(y_test_rev, y_pred_rev, target_names=['Negative', 'Positive'])
print(report)

# Visualize Confusion Matrix
cm_rev = confusion_matrix(y_test_rev, y_pred_rev)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_rev, annot=True, fmt='d', cmap='Greens',
            xticklabels=['Negative', 'Positive'],
            yticklabels=['Negative', 'Positive'])
plt.title('Confusion Matrix - Sentiment Analysis')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()

# --- Deliverable 4: Insights ---
print("\n--- Insights ---")
# Examples of reviews classified correctly and incorrectly
# (Since the test set is small, we'll just show the whole test set's predictions)
df_test_results = df_reviews.loc[y_test_rev.index]
df_test_results['Predicted_Sentiment'] = ['positive' if p == 1 else 'negative' for p in y_pred_rev]
print("\nPrediction Results on Test Data:")
print(df_test_results[['Review_Text', 'Sentiment', 'Predicted_Sentiment']])

# Show common features (words) for positive and negative reviews
feature_names = tfidf_vectorizer.get_feature_names_out()
coefficients = sentiment_model.coef_[0]

# Create a DataFrame of words and their coefficients
coef_df = pd.DataFrame({'Word': feature_names, 'Coefficient': coefficients})

# Top words for positive sentiment (highest coefficients)
top_positive_words = coef_df.sort_values(by='Coefficient', ascending=False).head(5)
print("\nTop 5 Words Associated with POSITIVE Sentiment:")
print(top_positive_words)

# Top words for negative sentiment (lowest coefficients)
top_negative_words = coef_df.sort_values(by='Coefficient', ascending=True).head(5)
print("\nTop 5 Words Associated with NEGATIVE Sentiment:")
print(top_negative_words)
```
